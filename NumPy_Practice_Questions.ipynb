### Question 12
Using a **boolean mask**, get the even numbers greater than 10 from the array.

**Expected output:**
```
[12 14 16 18 20]
```

````python
x = np.arange(1, 21)

# Write your solution here

# Note: In NumPy, use '&' instead of 'and' for element-wise comparisons,
#  and wrap each condition in ()
mask =( (x%2 == 0) & (x>10) )
filtered_arr = x[mask]
print(filtered_arr)
````

## Level 3: Reshaping and Combining Arrays

### Question 14
Reshape `np.arange(12)` into shape (3, 4), then (2, 6), then (4, -1). What does `-1` mean?

````python
# Write your solution 

arr = np.arange(12)
reshaped_arr1 = np.reshape(arr,(3,4))
print("Shape (3,4): \n",reshaped_arr1)
reshaped_arr2 = np.reshape(arr,(2,6))
print("Shape (2,6): \n",reshaped_arr2)

# (-1) automatically calculates the dimension size. 
reshaped_arr3 = np.reshape(arr,(4,-1))
print("Shape (4,-1): \n",reshaped_arr3)
````

### Question 15
Convert the matrix to 1D using `flatten()` and then `ravel()`. What is the difference? (Try modifying the result and check whether the original changes.)

````python
m = np.array([[1, 2, 3], [4, 5, 6]])

# Write your solution here

# flatten(): Returns a copy of the array as a 1D array.
# ravel(): Returns a view of the array as a 1D array whenever possible.

arr2 = m.flatten()
print("1D array using flatten(): \n",arr2)
arr2[0]=26
print("original array after modifiying one element in 1D array made by flatten(): \n",m)
arr3 = m.ravel()
print("1D array using ravel(): \n",arr3)
arr3[0]=26
print("original array after modifiying one element in 1D array made by ravel(): \n",m)
````

### Question 17
Combine `a` and `b`:
1. Vertically
2. Horizontally

Try it with `np.concatenate`, then with `vstack` / `hstack`.

````python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

# Write your solution here

# concatenate(): Joins arrays along a specified existing axis.
# vstack(): Stacks arrays vertically (row-wise).
# hstack(): Stacks arrays horizontally (column-wise).

vertical_concat = np.concatenate((a,b),axis=0)
horizontal_concat = np.concatenate((a,b),axis=1)
print("concat with axis=0 :\n",vertical_concat)
print("concat with axis=1 :\n",horizontal_concat)

v_stack = np.vstack((a,b))
h_stack = np.hstack((a,b))
print("vertical stack : \n",v_stack)
print("horizontal stack : \n",h_stack)
````
